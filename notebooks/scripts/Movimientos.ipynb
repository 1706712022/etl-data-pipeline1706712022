````python
import pandas as pd
````

````python
url = "https://raw.githubusercontent.com/1706712022/etl-data-pipeline1706712022/refs/heads/main/data/raw/E_movimientos.csv"
````

````python
df_mov = pd.read_csv(url)
print("Dataset Cargado Correctamente")
````

````python
df_mov.head()
````

````python
print ("Primeras filas del Dataset");
df_mov.head()
````

````python
print("Data Types and Non-Null Counts:")
df_mov.info()

print("\nMissing Values per Column:")
print(df_mov.isnull().sum())

print("\nDescriptive Statistics:")
df_mov.describe()
````

````python
df_mov.columns = [col.lower().replace(' ', '_') for col in df_mov.columns]
print("Column names standardized:")
df_mov.head()
````

````python
df_mov['cantidad_movimiento'] = df_mov['cantidad_movimiento'].astype(str).str.replace(r'\D', '', regex=True)
df_mov['cantidad_movimiento'] = pd.to_numeric(df_mov['cantidad_movimiento'], errors='coerce')

print("Data Types after cleaning 'cantidad_movimiento':")
df_mov.info()

print("\nFirst few rows after cleaning 'cantidad_movimiento':")
df_mov.head()
````

````python
rejects_df_mov = df_mov[df_mov['cantidad_movimiento'].isna()].copy()
curated_df_mov = df_mov[df_mov['cantidad_movimiento'].notna()].copy()

print(f"Shape of rejects_df_mov: {rejects_df_mov.shape}")
print(f"Shape of curated_df_mov: {curated_df_mov.shape}")
````

````python
print("df_mov - Primeras 5 filas:")
df_mov.head()

print("\ndf_mov - Información:")
df_mov.info()

print("\ncurated_df_mov - Primeras 5 filas:")
curated_df_mov.head()

print("\ncurated_df_mov - Información:")
curated_df_mov.info()

print("\nrejects_df_mov - Primeras 5 filas:")
rejects_df_mov.head()

print("\nrejects_df_mov - Información:")
rejects_df_mov.info()
````

````python
curated_df_mov['fecha'] = pd.to_datetime(curated_df_mov['fecha'], errors='coerce')

print("curated_df_mov - Data types after converting 'fecha' column:")
curated_df_mov.info()

print("\ncurated_df_mov - Primeras 5 filas después de convertir 'fecha':")
curated_df_mov.head()
````

````python
print("\nMissing values in curated_df_mov before dropping:")
print(curated_df_mov[['id_inventario', 'fecha']].isnull().sum())

curated_df_mov.dropna(subset=['id_inventario', 'fecha'], inplace=True)

print("\nShape of curated_df_mov after dropping rows with missing 'id_inventario' or 'fecha':", curated_df_mov.shape)
print("\ncurated_df_mov - Information after dropping nulls:")
curated_df_mov.info()
````

````python
print("rejects_df_mov - Todas las filas para inspección:")
print(rejects_df_mov)
````
